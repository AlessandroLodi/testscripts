```python
import os
import numpy as numpy
import matplotlib.pyplot as plt
import string
from imports.qtlab_data import *
from imports.dataclass import *
from imports.physics_models import *
from matplotlib.colors import to_rgba
```

```python
## Change the directory and Configure Import Settings

os.chdir(r"G:\2021\AG_LG05_3")
chipPiece           = "AG_LG05_3_GNR_anthracene"
folder_electroburn  = "mol gnr_anthracene_dil1to100_tol"
folder_molps        = "mol gnr_anthracene_r_gs_vsd100mV"
folder_molps_2      = "mol gnr_anthracene_r_gs_vsd100mV"
folder_molps_3      = "mol gnr_methoxy_dil1to100_tol_gs_sd_goodev"
folder_molps_4      = "mol gnr_methoxy_dil1to100_tol_gs_sd_goodev_2ndTest"


## Import Data

match   = "{}|{}|{}|{}|{}".format(folder_molps_4, folder_molps_3, folder_molps_2, folder_molps, folder_electroburn)
pattern = ".*?(?P<folder>{}).*?\d+_(?P<exp>[a-zA-Z0-9_]+)_(?P<type>[a-zA-Z0-9\-_]+?)_(?P<device>[a-zA-Z]+[0-9]+)\.(?:dat|csv|txt)".format(match)
dset    = QTLab_Dataset.find(pattern=pattern)
dset    = dset[np.argsort(dset["timestamp"])[::-1]]

## Slice Datset

ivsvgset = dset[dset["type"] == "IVsVg"]
ivsvgset_electroburn    = ivsvgset[ivsvgset["folder"] == folder_electroburn]
ivsvgset_molps          = ivsvgset[ivsvgset["folder"] == folder_molps]
ivsvgset_molps_2        = ivsvgset[ivsvgset["folder"] == folder_molps_2]
ivsvgset_molps_3        = ivsvgset[ivsvgset["folder"] == folder_molps_3]
ivsvgset_molps_4        = ivsvgset[ivsvgset["folder"] == folder_molps_4]


devs = np.unique(ivsvgset_electroburn['device'])

for i in devs:
    current_list    = []
    fig             = Figure()
    d               = ivsvgset_electroburn[ivsvgset_electroburn['device'] == i]
    data            = d.load(Stability_Diagram)
    data_mol_container = ivsvgset_molps[ivsvgset_molps['device'] == i]
    data_mol           = data_mol_container.load(Stability_Diagram)
    data_mol_container2 = ivsvgset_molps_2[ivsvgset_molps_2['device'] == i]
    data_mol2           = data_mol_container2.load(Stability_Diagram)
    data_mol_container3 = ivsvgset_molps_3[ivsvgset_molps_3['device'] == i]
    data_mol3           = data_mol_container3.load(Stability_Diagram)
    data_mol_container4 = ivsvgset_molps_4[ivsvgset_molps_4['device'] == i]
    data_mol4           = data_mol_container4.load(Stability_Diagram)
    if data:
        data = data[0]
        data.resample(256, 256)
        current_list.append(np.mean(np.abs(data['Isd'].values)) * 5)
        fig.add_subplot(Subplot_IVsVg(data, title='No GNR', crange=(-np.max(current_list), np.max(current_list)), cmap='viridis'));
        #############
    if data_mol:
        data_mol = data_mol[0]
        data_mol.resample(256, 256)
        current_list.append(np.mean(np.abs(data_mol['Isd'].values)) * 5)
        fig.add_subplot(Subplot_IVsVg(data_mol, title='No GNR', crange=(-np.max(current_list), np.max(current_list)), cmap='viridis'));
        #############
    if data_mol2:
        data_mol2 = data_mol2[0]
        data_mol2 = data_mol2.resample(256, 256)
        current_list.append(np.mean(np.abs(data_mol2['Isd'].values)) * 5)
        fig.add_subplot(Subplot_IVsVg(data_mol2, title='2nd Check', crange=(-np.max(current_list), np.max(current_list)), cmap='viridis'));
        #############
    if data_mol3:
        data_mol3 = data_mol3[0]
        data_mol3 = data_mol3.resample(256, 256)
        current_list.append(np.mean(np.abs(data_mol3['Isd'].values)) * 5)
        fig.add_subplot(Subplot_IVsVg(data_mol3, title='3rd Check', crange=(-np.max(current_list), np.max(current_list)), cmap='viridis'));
        #############
    if data_mol4:
        data_mol4 = data_mol4[0]
        data_mol4 = data_mol4.resample(256, 256)
        current_list.append(np.mean(np.abs(data_mol4['Isd'].values)) * 5)
        fig.add_subplot(Subplot_IVsVg(data_mol4, title='4th Check', crange=(-np.max(current_list), np.max(current_list)), cmap='viridis'));
    

    fig.visualise(f'figure_nb_same_colorbar_SD_all/{chipPiece}/{i}')




```
