```
class Book:
    BOOK_TYPES = ("HARDCOVER", "PAPERBACK", "EBOOK")

    # keep track of the book list
    __booklist = None

    def __init__(self, title, author, pages, price, booktype):
        self.author = author
        self.pages = pages
        self.price = price
        self.title = title # the attribute is create for hold the title
        self.__secret = 'This is a secret attribut'
        if (not booktype in Book.BOOK_TYPES):
            raise ValueError(f"{booktype} is not a valid book type")
        else:
            self.booktype = booktype

    def getprice(self):
        # becuause was not define in the init. So if is there, use it.
        if hasattr(self, "_discount"):
            return self.price - (self.price * self._discount)
        else:
            return self.price
    
    def setdiscount(self, amount):
        self._discount = amount

    @classmethod # work on the entire class
    def get_booktype(cls):
        return cls.BOOK_TYPES
    
    @staticmethod
    def get_booklist():
        if Book.__booklist == None:
            Book.__booklist = []
        return Book.__booklist

```

```
b2 = Book("War and Piece", "Leo Tolstoy", 1225, 39.95, "HARDCOVER")
print(b2)
print(b2.getprice())
```

```
from abc import ABC, abstractmethod

class JSONify(ABC):
    @abstractmethod
    def toJSON(self):
        pass


class GraphicShape(ABC):
    def __init__(self):
        super().__init__()
    
    @abstractmethod
    def calcArea(self):
        pass

class Circle(GraphicShape, JSONify):
    def __init__(self, radius):
        self.radius = radius
    
    def calcArea(self):
        return 3.14 * self.radius ** 2
    
    def toJSON(self):
        return f"{{\" circle\" : {str(self.calcArea())} }}"

class Square(GraphicShape):
    def __init__(self, side):
        self.side = side
    
    def calcArea(self):
        return self.side ** 2

c = Circle(10)
print(c.calcArea())
print(c.toJSON())

s = Square(10)
print(s.calcArea())
# g = GraphicShape()



```
